# Lab 3. Exercise 2.C: Working with tax functions

```python
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062
```

```python
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145
```

```python

def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:
        rate = 0.185
    else:
        rate = 0.18 - ((withholding_code - 4) * 0.005)

    return gross_pay * rate
```

```python
people = [
    (750, 0),
    (1550, 2),
    (1100, 5)
]
```

```python
for gross, code in people:
    print(f"Gross Pay: ${gross}")
    print(f"Social Security Tax: ${get_soc_sec_tax(gross):.2f}")
    print(f"Medicare Tax: ${get_medicare_tax(gross):.2f}")
    print(f"Federal Tax: ${get_federal_tax(gross, code):.2f}")
    print()
```
