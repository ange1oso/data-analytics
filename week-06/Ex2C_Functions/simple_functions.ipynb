#  Lab 1. Exercise 2.C: Working with defining custom functions

```python
def favorite_things(name, movie, music):
    return f"Name: {name}\nFavorite Movie: {movie}\nFavorite Musician: {music}"

print(favorite_things("Angel", "Sicario", "Drake"))
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined the Data Analyst program to transition into a more rewarding career."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place}, because {why}."

print(favorite_place("Chicago", "there is always something to do"))
```

```python
print(favorite_place("Mexico", "the beaches and weather are beautiful"))
```
