# Exercise 2.A Using built-in functions 
## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(part_code):
    return part_code.split(':')[0]
```

```python
def get_product_num(part_code):
    return part_code.split(':')[1].split('-')[0]
```

```python
def get_size(part_code):
    return part_code.split('-')[1]
```

```python
part1 = 'ACME:123-L'
part2 = 'DI:12345-M'
part3 = 'ACE:1-12'
```

```python
for part in [part1, part2, part3]:
    supplier = get_supplier_code(part)
    number = get_product_num(part)
    size = get_size(part)

    print(f"For {part}, the supplier is {supplier}. Part number {number} is available in size {size}.")
```
