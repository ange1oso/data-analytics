# Exercise 3.A Getting Familiar with Matplotlib

```python

import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python

x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python

plt.plot(x, y)
plt.show()
```

```python

plt.plot(x, y, marker='o')
plt.show()
```

```python

plt.plot(x, y, marker='o')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

## Styling Plots

```python

plt.plot(x, y, linestyle='--', color='red')
plt.show()
```

```python

plt.plot(x, y, marker='s', markersize=10, markerfacecolor='green')
plt.show()
```

```python

plt.plot(x, y, marker='^', color='purple')
plt.show()

plt.plot(x, y, marker='d', color='orange')
plt.show()
```

# Exercise 3.B More Plotting Practice

## Advanced Plotting Techniques

```python

products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products, sales, color='skyblue', width=0.5)
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python

x = np.random.rand(50)
y = np.random.rand(50)
sizes = np.random.rand(50) * 100
colors = np.random.rand(50)

plt.scatter(x, y, s=sizes, c=colors, cmap='viridis')
plt.title('Scatter Plot')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.show()
```

## Subplots

```python

fig, axs = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0, 10, 100)
y = np.sin(x)

axs[0].plot(x, y)
axs[0].set_title('Sine Wave')
axs[0].set_xlabel('X')
axs[0].set_ylabel('sin(x)')

axs[1].hist(np.random.randn(1000))
axs[1].set_title('Histogram')
axs[1].set_xlabel('Value')
axs[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
