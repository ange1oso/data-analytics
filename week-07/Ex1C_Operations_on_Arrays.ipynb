# Exercise 1.C Performing Operations on Arrays

```python

import numpy as np
```

## Sample Arrays

```python

r_1 = np.random.randint(1, 20, size=2)
r_2 = np.random.randint(1, 20, size=(1, 2))
r_3 = np.random.rand(2, 1)
r_4 = np.random.randint(20, 100, size=(4, 5))
r_5 = np.random.rand(4, 5)
```

## Permutations

```python

print(r_4)
```

```python

np.random.shuffle(r_4)
print(r_4)
```

shuffle changes the original array order directly.

```python

new_shuffle = np.random.permutation(r_4)

print(new_shuffle)
print(r_4)
```

shuffle modifies the original array, while permutation returns a shuffled copy.

## Mathematical Operations

```python

print(r_1, r_1.shape)
print(r_2, r_2.shape)

result = np.add(r_1, r_2)
print(result)
print(result.shape)
```

```python

print(r_1, r_1.shape)
print(r_3, r_3.shape)

result = np.add(r_1, r_3)
print(result)
print(result.shape)
```

```python

print(r_2, r_2.shape)
print(r_4, r_4.shape)

result = np.add(r_2, r_4)
print(result)
print(result.shape)
```

```python

# This produces a broadcasting error because the shapes are incompatible.

# np.add(r_1, r_4)

print(np.add(r_2, r_4))
print(np.add(r_3, r_4))
```

```python

print(np.add(np.array([1]), r_4))
```

```python

print(np.add(np.array([[1]]), r_4))
```

```python

print(np.add(np.array([1,1,1,1,1]), r_4))
```

```python

# This creates a broadcasting error because the dimensions do not align.

# print(np.add(np.array([[1],[1]]), r_4))
```


Broadcasting allows NumPy to perform operations on arrays with different shapes.
Smaller arrays are expanded automatically when dimensions are compatible.
The resulting shape is based on the larger compatible array.

## Grocery Problem

```python

bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])

print(bag1)
print(bag2)
```

```python

print(f"The first bag contains {np.sum(bag1)} items.")
print(f"The second bag contains {np.sum(bag2)} items.")
```

```python

totals = np.add(bag1, bag2)

print(f"There are {totals[0]} vegetables, {totals[1]} fruits, "
      f"{totals[2]} boxes of snacks, and {totals[3]} beverages.")
```

## Teamwork

```python

ali = np.array([5, 4, 6, 4, 2])
ben = np.array([0, 5, 5, 3, 4])
carla = np.array([3, 6, 0, 3, 4])

all_knocks = np.array([ali, ben, carla])
print(all_knocks)
```

```python

person_totals = np.sum(all_knocks, axis=1)
group_total = np.sum(all_knocks)

print(person_totals)
print(group_total)
```

```python

daily_totals = np.sum(all_knocks, axis=0)
rolling_total = np.cumsum(daily_totals)
estimated_votes = rolling_total * 0.5

print(estimated_votes)
```
