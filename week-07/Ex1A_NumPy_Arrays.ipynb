# Exercise 1.A NumPy: Manipulating Arrays 

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python

for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop works across the rows of the array, so entire rows are multiplied at once.
The second loop works through each individual value inside the rows.

```python

for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# This result is different because x represents the entire row array,
# not the individual value y.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA))
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```


This produces an error because the arrays have different shapes.
joined_array_1 has shape (4,3) while joined_array_2 has shape (2,6)
stacked_array_2 = np.stack((joined_array_1, joined_array_2))


```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```


This creates an error because the arrays do not have the same
number of rows when concatenating on axis=1.
another_2 = np.concatenate((joined_array_1, arrA), axis=1)


```python

horizontal_stack = np.hstack((arrA, arrA))
# hstack joins arrays horizontally by columns
print(horizontal_stack)

vertical_stack = np.vstack((arrA, arrA))
# vstack joins arrays vertically by rows
print(vertical_stack)

depth_stack = np.dstack((arrA, arrA))
# dstack stacks arrays by depth
print(depth_stack)
```

## Splittin Arrays

```python

splitsky = np.array_split(joined_array_1, 2)
print(splitsky)

# Result: 2 arrays, each with 2 dimensions.
```

```python

spliterino = np.array_split(joined_array_1, 3)
print(spliterino)

# Result: 3 arrays with uneven row distribution.
```

## Searching and Replacing Array Values

```python

result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python

arrB = np.array([[7, 6], [13, 5], [7, 3]])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python

print(arrB)

z = arrB < 7
arrC = arrB[z]

print(arrC)
```

```python

y = arrB >= 7
arrD = arrB[y]

print(arrD)
```
